```python
from practical_ml_ai.language import MODEL_ID, MODEL_REVISION, redact_contacts
example = split["train"].iloc[0]
print("Actual base model/revision:", MODEL_ID, MODEL_REVISION)
print("Training example ID:", example.text_id)
print("Length before/after declared masking:",
      len(example.text), len(redact_contacts(example.text)))
# The content remains local; masking is not a guarantee that all personal data vanish.
```
