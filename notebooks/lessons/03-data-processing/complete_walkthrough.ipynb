## NumPy arrays and a small linear system

The request counts below are a separate illustration from latency. Inspect shape and units. Then solve: two A plus one B requires eight resource units; one A plus two B requires seven. Verify by substitution.


```python
requests = np.array([12, 18, 20])
print("Request-count array shape:", requests.shape)
print("Total requests:", requests.sum())
print("Mean requests:", requests.mean())
coefficients = np.array([[2, 1], [1, 2]])
requirements = np.array([8, 7])
solution = np.linalg.solve(coefficients, requirements)
print("Resource units for A and B:", solution)
print("Substituted requirements:", coefficients @ solution)
```
