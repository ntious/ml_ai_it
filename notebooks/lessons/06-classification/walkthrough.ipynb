## Predict and inspect the mechanism

A probability estimate is not automatically calibrated confidence. A threshold converts a score into a decision. False positives and false negatives can have different costs; balanced accuracy gives equal weight to recall in each class, not to operational cost.


```python
# An actual training rating becomes a declared classroom label.
row = split["train"].iloc[0]
print("Observed rating:", int(row.quality))
print("Declared positive class (rating >= 7):", int(row.quality >= 7))
print("This threshold is a task definition, not a certification.")
```
