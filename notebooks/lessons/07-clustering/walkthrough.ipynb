## Predict and inspect the mechanism

Two columns measured on very different scales can dominate Euclidean distance differently. Centering and scaling use training means/standard deviations. Compare original-unit profile means after fitting; do not interpret the standardized number as a physical measurement.

K-means minimizes squared distances to centers, so units and scaling affect the result. Silhouette contrasts within-cluster closeness with separation from other clusters. It is not classification accuracy. Cluster IDs can be permuted without changing the partition.


```python
from sklearn.preprocessing import StandardScaler
# Demonstrate fitting scope on real training measurements only.
scaler_demo = StandardScaler().fit(split["train"][features])
print("First three training centers:", scaler_demo.mean_[:3])
print("First three training scales:", scaler_demo.scale_[:3])
print("Quality is excluded from the feature list:", "quality" not in features)
```

## Guided construction before independent practice

Without scaling, a feature measured in thousands can dominate a feature measured in fractions. K-means minimizes squared distance to centers. Its IDs have no inherent order or meaning.

Predict the output before running. Then complete the module guided page's independent modification.


```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from practical_ml_ai.continuation_data import wine_splits
learning_split, learning_features, _ = wine_splits()
learning_scaler = StandardScaler().fit(learning_split["train"][learning_features])
learning_x = learning_scaler.transform(learning_split["train"][learning_features])
learning_clusterer = KMeans(n_clusters=2, n_init=10, random_state=42).fit(learning_x)
print("centers in original units", learning_scaler.inverse_transform(learning_clusterer.cluster_centers_))
assert learning_clusterer.cluster_centers_.shape == (2,len(learning_features))
```
