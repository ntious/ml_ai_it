| Message | Existing evidence | What to do |

| Final evaluation already completed | `evaluation.json` | Read the saved final results rather than rerunning evaluation on this folder. |

These errors commonly occur when an individual stage cell is executed twice for the same `case_run`. They preserve the recorded experiment; they do not mean training failed.
