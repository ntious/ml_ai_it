### route_intent
Return count_rows for count, mean_alcohol for alcohol; otherwise None.

Visible input arguments: `['count']`. Predict the result before running.

Hint: An unsupported intent must not fall through to an unrelated answer.

```python
def route_intent(intent):
    return {"count":"count_rows","alcohol":"mean_alcohol"}.get(intent)
```

### allowed_call
Allow only count_rows with no arguments, or mean_alcohol with no arguments.

Visible input arguments: `['count_rows', {}]`. Predict the result before running.

Hint: Validate tool and argument contract, not just words in a prompt.

```python
def allowed_call(tool, arguments):
    return tool in {"count_rows","mean_alcohol"} and arguments == {}
```

```python
results['m14_allowed_call'] = check('m14_allowed_call', allowed_call)
```

### can_call
Return True for nonnegative whole-number used/budget and used < budget; reject bool.

Visible input arguments: `[0, 1]`. Predict the result before running.

Hint: Stop before exceeding the budget and reject malformed limits.

```python
def can_call(used, budget):
    return type(used) is int and type(budget) is int and 0 <= used < budget
```

```python
results['m14_can_call'] = check('m14_can_call', can_call)
```
