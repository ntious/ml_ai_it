### error_counts
Return [false_alerts, misses] for equally sized boolean lists.

Try the call `error_counts([False, True, True], [True, False, True])`. Predict the result before running.

Hint: Count errors by their consequences, not only total accuracy.

```python
def error_counts(actual, predicted):
    return [sum((not a) and p for a,p in zip(actual,predicted)), sum(a and not p for a,p in zip(actual,predicted))]
```

```python
results['m01_error_counts'] = check('m01_error_counts', error_counts)
```
