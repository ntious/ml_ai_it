### parse_label
Return ham/spam only for an exact one-key JSON object; otherwise return None.

Try the call `parse_label('{"label":"spam"}')`. Predict the result before running.

Hint: Reject prose and extra keys; parsing is separate from correctness.

```python
def parse_label(text):
    import json
    try:
        value = json.loads(text)
    except (ValueError,TypeError):
        return None
    return value["label"] if isinstance(value,dict) and set(value)=={"label"} and value["label"] in ["ham","spam"] else None
```

```python
results['m10_parse_label'] = check('m10_parse_label', parse_label)
```

### valid_fraction
Input parsed labels or None. Return valid/total, or None for empty input.

Try the call `valid_fraction(['spam', None, 'ham'])`. Predict the result before running.

Hint: Count malformed outputs in the denominator.

```python
def valid_fraction(labels):
    return sum(x in ["ham","spam"] for x in labels)/len(labels) if labels else None
```

```python
results['m10_valid_fraction'] = check('m10_valid_fraction', valid_fraction)
```

### end_to_end_accuracy
Return exact matches / all cases, counting None as wrong; None for no cases.

Try the call `end_to_end_accuracy(['spam', 'ham'], ['spam', None])`. Predict the result before running.

Hint: Do not condition accuracy on the outputs the parser accepted.

```python
def end_to_end_accuracy(actual, parsed):
    return sum(a==p for a,p in zip(actual,parsed))/len(actual) if actual else None
```

```python
results['m10_end_to_end_accuracy'] = check('m10_end_to_end_accuracy', end_to_end_accuracy)
```
