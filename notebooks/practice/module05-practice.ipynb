```python
demo_verified = False
if len(results) == 3 and all(item["state"] == "passed" for item in results.values()):
    from practical_ml_ai.regression import download_bike_data
    from sklearn.dummy import DummyRegressor
    path = download_bike_data(ROOT / "runs/datasets/bike-sharing")
    # Development sandbox: first 81 source dates, not the complete study's final split.
    daily = pd.read_csv(path).iloc[:81].copy()
    daily["prior_day"] = daily.cnt.shift(1)
    daily["prior_week"] = daily.cnt.shift(7)
    development = daily.dropna(subset=["prior_day", "prior_week"]).copy()
    training, validation = development.iloc[:60], development.iloc[60:]
    features = ["prior_day", "prior_week"]
    model, prediction = fit_regression(training[features], training.cnt, validation[features])
    dummy = DummyRegressor().fit(training[features], training.cnt)
    table = pd.DataFrame({"date": validation.dteday, "actual_rentals": validation.cnt,
                          "linear": prediction, "previous_day": validation.prior_day,
                          "training_mean": dummy.predict(validation[features])}).reset_index(drop=True)
    comparison = pd.DataFrame([{"method": name, "validation_MAE_rentals": residual_summary(table.actual_rentals, table[name])["mae"]}
                               for name in ["linear", "previous_day", "training_mean"]])
    display(comparison.round(2), table.round(2))
    ax = table.set_index("date")[["actual_rentals", "linear", "previous_day"]].plot(
        figsize=(9, 4), style=["o-", "s--", "^:"], color=["#303030", "#17658b", "#a65c00"])
    ax.set(title="Development sandbox: 14 rolling one-day forecasts", xlabel="Target date (2011)", ylabel="Rentals per day")
    positions = [0, 3, 6, 9, len(table)-1]
    ax.set_xticks(positions)
    ax.set_xticklabels(table.date.iloc[positions], rotation=35, ha="right")
    plt.tight_layout(); plt.show()
    demo_verified = len(training) == 60 and len(validation) == 14 and len(table) == 14
    print("Source: UCI Bike Sharing, DOI 10.24432/C5W894; cached source digests verified by the downloader.")
    print("Known prior-day outcomes become available at each new origin. This is not a batch forecast issued before all 14 days.")
else:
    print("Development sandbox not run: finish the three functions and rerun from Setup.")
```
