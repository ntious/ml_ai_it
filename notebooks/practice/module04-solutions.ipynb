### median
Return the median of a nonempty numerical list, without changing the input.

Try the call `median([1, 2, 3, 100])`. Predict the result before running.

Hint: Average the two middle observations for an even count.

```python
def median(values):
    ordered = sorted(values)
    n = len(ordered)
    return ordered[n//2] if n%2 else (ordered[n//2-1]+ordered[n//2])/2
```

```python
results['m04_median'] = check('m04_median', median)
```

### weighted_rate
Input pairs [errors, requests]. Return total errors / total requests, or None.

Try the call `weighted_rate([[1, 2], [1, 8]])`. Predict the result before running.

Hint: Do not average daily percentages with unequal denominators.

```python
def weighted_rate(counts):
    denominator = sum(row[1] for row in counts)
    return sum(row[0] for row in counts)/denominator if denominator else None
```

```python
results['m04_weighted_rate'] = check('m04_weighted_rate', weighted_rate)
```
