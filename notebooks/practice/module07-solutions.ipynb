### squared_distance
Return sum of squared coordinate differences for equal-length vectors.

Try the call `squared_distance([2, 4], [1, 2])`. Predict the result before running.

Hint: Square coordinate differences; do not compare signed sums.

```python
def squared_distance(point, center):
    return sum((a-b)**2 for a,b in zip(point,center))
```

### select_k
Input pairs [K, validation_silhouette]; maximize score, smallest K breaks ties.

Try the call `select_k([[2, 0.4], [3, 0.5], [4, 0.45]])`. Predict the result before running.

Hint: Use validation geometry and a declared tie-break, not smallest inertia.

```python
def select_k(candidates):
    return min(candidates, key=lambda row:(-row[1],row[0]))[0]
```

```python
results['m07_select_k'] = check('m07_select_k', select_k)
```

### profile
Return string cluster IDs mapped to count and mean of a single original-unit feature.

Try the call `profile([2, 4, 10], [0, 0, 1])`. Predict the result before running.

Hint: Describe both support and feature level; IDs have no rank.

```python
def profile(values, labels):
    result = {}
    for label in sorted(set(labels)):
        subset = [v for v,k in zip(values,labels) if k == label]
        result[str(label)] = {"count":len(subset),"mean":sum(subset)/len(subset)}
    return result
```

```python
results['m07_profile'] = check('m07_profile', profile)
```
