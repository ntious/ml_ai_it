### status_counts
Return a dictionary counting the status string in each record.

Try the call `status_counts([{'status': 'open'}, {'status': 'closed'}, {'status': 'open'}])`. Predict the result before running.

Hint: Initialize unseen statuses and handle an empty file.

```python
def status_counts(records):
    counts = {}
    for row in records:
        counts[row["status"]] = counts.get(row["status"], 0) + 1
    return counts
```

```python
results['m02_status_counts'] = check('m02_status_counts', status_counts)
```

### observed_mean
Return the mean excluding None; return None if no values are observed.

Try the call `observed_mean([10, None, 20])`. Predict the result before running.

Hint: Missing is not zero; zero is still observed.

```python
def observed_mean(values):
    observed = [x for x in values if x is not None]
    return sum(observed)/len(observed) if observed else None
```

```python
results['m02_observed_mean'] = check('m02_observed_mean', observed_mean)
```

### output_path
Return the string path root/runs/run_name/summary.json using pathlib.

Try the call `output_path('course', 'trial-2')`. Predict the result before running.

Hint: Join root, runs, the run name and summary.json; keep outputs separate from source files.

```python
def output_path(root, run_name):
    from pathlib import Path
    return (Path(root)/"runs"/run_name/"summary.json").as_posix()
```

```python
results['m02_output_path'] = check('m02_output_path', output_path)
```
