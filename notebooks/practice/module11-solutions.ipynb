### trainable_count
Input [size, trainable] pairs; return number of trainable scalar values.

Try the call `trainable_count([[100, False], [8, True], [2, True]])`. Predict the result before running.

Hint: Count scalars in trainable tensors, not the number of tensors.

```python
def trainable_count(parameters):
    return sum(size for size,trainable in parameters if trainable)
```

```python
results['m11_trainable_count'] = check('m11_trainable_count', trainable_count)
```

### completion_labels
Return ids with prompt and padding positions replaced by -100.

Try the call `completion_labels([5, 6, 7, 0], 2, [1, 1, 1, 0])`. Predict the result before running.

Hint: Mask both prompt and padding, while retaining completion tokens.

```python
def completion_labels(ids, prompt_length, attention_mask):
    return [token if i>=prompt_length and attention_mask[i] else -100 for i,token in enumerate(ids)]
```

```python
results['m11_completion_labels'] = check('m11_completion_labels', completion_labels)
```

### changed_tensors
Input equally sized lists of scalar lists. Return indices of tensors with any changed value.

Try the call `changed_tensors([[1, 2], [3]], [[1, 2], [4]])`. Predict the result before running.

Hint: Demonstrate an actual parameter change; a new loss string is insufficient.

```python
def changed_tensors(before, after):
    return [i for i,(a,b) in enumerate(zip(before,after)) if a != b]
```

```python
results['m11_changed_tensors'] = check('m11_changed_tensors', changed_tensors)
```
